#### 2.2 Creating a Simulating Dataset

Problem: You need to generate a dataset of simulated data

scikit-learn offers many methods for creating simulated data. Of those, three methods
are particularly useful.
When we want a dataset designed to be used with linear regression, make_regression
is a good choice

Section 2.2 does not load an existing dataset. It generates simulated data with scikit-learn:

make_regression creates inputs and continuous outputs for regression.
make_classification creates inputs and category labels for classification.

```python
# Load library 

from sklearn.datasets import make_regression

# Generates feature matrix, target vector, and the true coefficients for a linear regression problem
features, target, coefficients = make_regression(n_samples=100, n_features=3, n_informative=3, n_targets=1, noise=0.0, coef=True, random_state=1)

# n_samples = 100: generate 100 samples
# n_features = 3: generate 3 features
# n_informative = 3: all 3 features are informative
# n_targets = 1: generate 1 target variable
# noise = 0.0: no noise in the data
# coef = True: return the true coefficients

#View feature matrix and target vector
print("Feature matrix:\n", features[:3])
print("Target vector:\n", target[:3])
```

```python
from sklearn.datasets import make_classification

# Generates feature matrix and target vector for a classification problem
features, target = make_classification(n_samples=100, n_features=3, n_informative=3, n_redundant=0, n_classes=2,weights=[.25, .75], random_state=1)

# n_samples = 100: generate 100 samples
# n_features = 3: generate 3 features
# n_informative = 3: all 3 features help separate the classes
# n_redundant = 0: no redundant features
# n_classes = 2: two classes to predict
# weights = [.25, .75]: 25% of the data belongs to the first class, 75% to the second

# View feature matrix and target vector
print("Feature matrix:\n", features[:3])
print("Target vector:\n", target[:3])
```

Finally, if we want a dataset designed to work well with clustering techniques, scikit
learn offers make_blobs:

make_blobs which generates groups of nearby points called blobs or clusters

```python
# Load library
from sklearn.datasets import make_blobs

# Generates feature matrix and target vector for clustering problem
features, target = make_blobs(n_samples=100, n_features=3, centers=3, cluster_std=0.5, shuffle=True, random_state=1)

# n_samples = 100: generate 100 samples
# n_features = 3: generate 3 features
# centers = 3: creates 3 seperate clusters
# cluster_std = 0.5: controls how spread out each cluster is, smaller values create tigher clusters
# shuffle = True: mixes the points up so that they are not ordered by cluster
# random_state = 1: ensures that the same random numbers are generated each time the code is run

#view feature matrix and target vector
print("Feature matrix:\n", features[:3])
print("Target vector:\n", target[:3])

# the labels are usually 0, 1, and 2 for the three clusters, but they can be any integer values.
# The important thing is that points with the same label belong to the same cluster.
```

Clustering is generally unsupervised learning, so a clustering algorithm normally receives only: features

make_blobs provides target mainly so you can compare the algorithm’s discovered clusters with the original, known cluster assignments. Therefore:

features are the generated points,
target contains the original cluster labels,
target is useful for evaluating a clustering algorithm, but is not normally given to the algorithm during training
